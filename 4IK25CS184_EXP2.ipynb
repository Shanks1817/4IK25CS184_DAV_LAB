```python
import pandas as pd
import numpy as np
```

```python
data = {'Name':["Amit","Bebo","Chetan","Amit","Divya","Ravi"],
        'Age':[25,np.nan,30,25,28,22],                                             # where np.nan represents missing values (nan = not a number)
        'Marks':[85,90,np.nan,75,95,80],
        'City':['Mysuru','Mysuru','Bengaluru','Mysuru',np.nan,'Delhi']}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of database (rows,columns):",df.shape)
print("\nColumn names:",df.columns.tolist())

print("\nData Type of each column:")
print(df.dtypes)

print("\nDataset info:")
df.info()
```

```python
print("\nDiscriptive statisctics (numerical columns):")
print(df.describe())

print("\nMean\n",df.mean(numeric_only=True))
print("\nMedian\n",df.median(numeric_only=True))
print("\nMode\n",df.mode(numeric_only=True))
print("\nStandard Deviation\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each columns:")
print(df.isnull().sum())

print("\nTotal missing values in dataset:", df.isnull().sum().sum())
```

```python
print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])

print("\nTotal duplicate rows in dataset:",df.duplicated().sum())
print("\nEDA process completed successfully!")
```
